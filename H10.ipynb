```python
# task2_count_strings.py
mixed = [42, "hello", 3.14, "world", True, None, "python", ["nested", "list"], {"a": 1}, ""]

count = 0
for item in mixed:
    if isinstance(item, str):
        count += 1

print(count)
```

```python
mylist = ["something important", 52, 3.14, "something extremely important", True, [1, 2, 3], {"a": 1}]
count_strings = 0

for element in mylist:
    if type(element) == str:
        count_strings += 1
print("Number of string elements:", count_strings)
```
